# Kesimpulan
Semua model mengalami peningkatan akurasi setelah dilakukan oversampling, baik menggunakan SMOTE maupun ADASYN.

Model Random Forest konsisten memberikan akurasi tertinggi pada data hasil oversampling, baik SMOTE (0.969) maupun ADASYN (0.939).

Sebelum oversampling, Bagging dengan Decision Tree memberikan performa terbaik (0.867).

Secara keseluruhan, Random Forest adalah model paling unggul untuk dataset ECOLI setelah dilakukan oversampling.
