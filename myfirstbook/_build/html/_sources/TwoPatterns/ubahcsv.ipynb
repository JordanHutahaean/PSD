```python
import pandas as pd

def ts_to_csv(ts_path, csv_path):
    data = []
    labels = []

    with open(ts_path, 'r') as file:
        for line in file:
            line = line.strip()

            # Lewati metadata
            if (
                not line or
                line.startswith('@') or
                line.startswith('#')
            ):
                continue

            # Pisahkan data dan label
            series, label = line.split(':')

            # Bersihkan tanda kurung
            series = series.replace('(', '').replace(')', '')

            # Konversi ke float
            values = list(map(float, series.split(',')))

            data.append(values)
            labels.append(label)

    # Buat DataFrame
    df = pd.DataFrame(data)
    df['label'] = labels

    # Simpan ke CSV
    df.to_csv(csv_path, index=False)
    print(f"Berhasil disimpan: {csv_path}")


# ======================
# CONTOH PEMAKAIAN
# ======================

ts_to_csv("TwoPatterns_TEST.ts", "TwoPatterns_TEST.csv")
```
